# Baseline Content-Based Recommendation

## Objective

```python
from culture_compass.recommender.content_based import (
    ContentBasedRecommender,
)
```

```python
df[
    df["event_name"]
    .str.contains(
        "abba",
        case=False,
        na=False,
    )
]
```
