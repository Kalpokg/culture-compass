```python
missing = (
    df.isnull()
      .sum()
      .sort_values(ascending=False)
)

missing
```

```python
print(
    "Duplicate event IDs:",
    df["event_id"].duplicated().sum()
)
```

```python
city_counts = df["city"].value_counts()
city_counts
```

```python
genre_counts = df["genre"].value_counts()
genre_counts
```

```python
segment_counts = (
    df["segment"]
      .value_counts()
)

segment_counts
```

```python
events_per_month = (
    df
    .groupby(df["date"].dt.to_period("M"))
    .size()
)

events_per_month
```

```python
df["text_length"] = (
    df["event_text"]
    .str.split()
    .str.len()
)

df["text_length"].describe()
```

```python
df[
    [
        "latitude",
        "longitude"
    ]
].describe()
```

# Imoportant Numbers

- Dataset contains 5139 unique events.

- No duplicate event IDs.

- Most events come from London. 

- The dominant genres are theatre, family,rock, comedy. Music events are less as they divided in subgroups such as rock, pop, jazz, classical, etc. 

- Missing values are minimal. Missing numbers - longitude and latitude:97,  genre:70, segment:13, time:6.

- Date range spans 07-2026 to 11-2027.
